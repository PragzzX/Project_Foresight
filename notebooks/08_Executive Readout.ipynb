# PROJECT FORESIGHT
## Executive Readout & Business Recommendations

### Inventory Intelligence, Demand Forecasting & Risk Management

---

**Project Objective**

Project FORESIGHT is an end-to-end inventory intelligence solution that
combines historical sales analysis, demand forecasting, inventory risk
scoring, financial impact estimation, and interactive decision-support
dashboards.

The system is designed to help business stakeholders:

- Forecast future product demand
- Identify potential stockout situations
- Detect excess inventory
- Prioritize replenishment decisions
- Quantify potential financial exposure
- Support data-driven inventory planning

---

### Solution Pipeline

**Raw Data → Data Pipeline → EDA → Feature Engineering → Forecasting
→ Risk Scoring → Dashboard → Prediction API**

---

### Executive Outcome

FORESIGHT transforms large-scale historical retail data into actionable
inventory recommendations, allowing stakeholders to move from
descriptive reporting toward proactive inventory management.

## 1. Executive Snapshot

Project FORESIGHT analysed **2,030,634 forecast records** and classified
inventory conditions into three operational risk categories.

| Overstock Risk | 593,552 |
| Healthy | 583,764 |

| Estimated Locked Capital | ₹9,475,885 |

### Key Executive Message

The analysis identifies significant opportunities to improve inventory

**Stockout risk** represents potential lost-sales exposure of
approximately **₹4.51 million**, while **overstock risk** represents
approximately **₹9.48 million of inventory capital that may be tied up**.

These findings provide a financial basis for prioritising inventory
actions rather than relying only on historical sales reports.

## 2. Business Problem

Retail inventory management requires balancing two competing risks:

### 1. Stockouts

Insufficient inventory can result in:

- Lost sales
- Lower customer satisfaction
- Missed demand
- Emergency replenishment
- Reduced revenue opportunities

### 2. Overstock

Excess inventory can result in:

- Capital being tied up
- Higher holding costs
- Slow-moving inventory
- Increased markdown requirements
- Reduced inventory efficiency

### Business Challenge

Traditional historical reporting can explain **what happened**, but
does not always provide sufficient support for deciding **what should
happen next**.

FORESIGHT addresses this gap by combining demand forecasting with
inventory risk classification and recommended actions.

## 3. Client / Business Context

FORESIGHT is designed around a large-scale retail inventory environment
where decisions must be made across:

- Multiple products
- Multiple stores
- Different demand patterns
- Seasonal periods
- Promotional activity
- Inventory constraints

At this scale, manually reviewing every SKU and store combination is
impractical.

The solution therefore focuses on **prioritisation**.

Instead of asking stakeholders to inspect millions of individual records,
FORESIGHT identifies records requiring attention and assigns an
appropriate business action.

### Decision Framework

| Risk | Priority | Recommended Decision |
|---|---|---|
| Stockout Risk | High | Reorder Immediately |
| Overstock Risk | Medium | Reduce Purchasing / Launch Promotion |
| Healthy | Low | Maintain Current Inventory |

## 4. Dataset & Data Pipeline

FORESIGHT was developed using large-scale retail sales data containing
product, store, calendar, pricing, and sales information.

The data pipeline was designed to transform raw extracts into an
analysis-ready dataset while keeping the process reproducible and
memory-efficient.

### Pipeline

**Raw Extracts**
→ Data Ingestion
→ Data Cleaning
→ Data Integration
→ Feature Engineering
→ Forecast Dataset
→ Risk Scoring Dataset
→ Dashboard & API

### Key Pipeline Decisions

- Missing values were handled programmatically.
- Duplicate records were identified and addressed.
- Data types were standardised.
- Sales data was reshaped from wide format into a long analytical
  structure.
- Calendar information was integrated with sales data.
- Feature engineering was performed for temporal and demand-related
  variables.
- Large datasets were processed in chunks to reduce memory pressure.

The final workflow was designed to be reproducible rather than relying
on manual data preparation.

## 5. Data Cleaning & Validation

Data quality was addressed before modelling to ensure that the
forecasting and risk-scoring stages operated on consistent inputs.

### Cleaning Areas

| Area | Treatment |
|---|---|
| Missing values | Identified and handled according to feature requirements |
| Duplicate records | Checked and removed where applicable |
| Data types | Standardised for analytical processing |
| Dates | Converted into usable temporal features |
| Sales structure | Reshaped into long format |
| Pricing | Integrated into modelling data |
| Calendar variables | Integrated with sales observations |

### Validation

The pipeline included validation checks after major processing stages
to ensure that:

- Required columns were available
- Data types were appropriate
- Integrated datasets remained consistent
- Forecast records were generated successfully
- Risk classifications were produced successfully

Large intermediate datasets were processed using chunk-based workflows
to avoid excessive memory consumption.

## 6. Exploratory Data Analysis

Exploratory analysis was performed to understand sales behaviour before
building the forecasting system.

### Key Areas Analysed

- Sales trends over time
- Weekly demand behaviour
- Product-level movement
- Store-level performance
- Regional sales patterns
- Seasonal behaviour
- Promotion and event effects
- Fast-moving and slow-moving inventory

### Regional Sales Insight

The analysis showed strong concentration in the highest-performing

The top-performing regions accounted for a substantial proportion of

### Business Interpretation

Demand is not evenly distributed across products, stores, or regions.

Therefore, applying a single inventory strategy across the entire
portfolio would be inefficient.

FORESIGHT addresses this by producing granular forecast and risk
information at the product/store level.

## 7. Forecasting Model

FORESIGHT uses a **Random Forest forecasting model** to estimate future
product demand.

### Model Features

The deployed forecasting model uses the following 10 features:

1. `sell_price`
2. `has_event`
3. `has_snap`
4. `is_weekend`
5. `lag_1`
6. `lag_2`
7. `lag_4`
8. `lag_8`
9. `rolling_mean_4`
10. `rolling_std_4`

### Feature Categories

**Price & Promotion Context**
- Selling price
- Event indicator
- SNAP indicator

**Calendar Context**
- Weekend indicator

**Historical Demand**
- 1-period lag
- 2-period lag
- 4-period lag
- 8-period lag

**Recent Demand Behaviour**
- 4-period rolling mean
- 4-period rolling standard deviation

### Forecasting Objective

The model provides a forward-looking demand estimate that can then be
used by the inventory risk engine to determine whether current
inventory levels require action.

## 8. Model Performance

The Random Forest forecasting model achieved the following reported
performance metrics:

| Metric | Result |
|---|---:|
| R² Score | 85.29% |
| MAE | 0.201 |
| RMSE | 0.534 |

### Interpretation

**R² = 85.29%**

The model explains approximately 85.29% of the variation represented by
the evaluated forecasting target.

**MAE = 0.201**

The mean absolute error provides an average magnitude of prediction
error of approximately 0.201 units in the evaluation scale.

**RMSE = 0.534**

The RMSE provides additional sensitivity to larger prediction errors.

### Business Interpretation

The model provides a strong forecasting foundation for the downstream
inventory-risk system, while the error metrics should continue to be
monitored as new data becomes available.

## 9. Inventory Risk Scoring

Forecasting alone does not tell a business what action to take.

FORESIGHT therefore converts forecast and inventory information into
three operational risk categories.

### Risk Framework

| Risk Category | Priority | Business Action |
|---|---|---|
| Stockout Risk | High | Reorder Immediately |
| Overstock Risk | Medium | Reduce Purchasing / Launch Promotion |
| Healthy | Low | Maintain Current Inventory |

### Risk Distribution

The validated risk-scoring dataset contains:

| Risk | Records |
|---|---:|
| Stockout Risk | 853,318 |
| Overstock Risk | 593,552 |
| Healthy | 583,764 |
| **Total** | **2,030,634** |

### Decision Logic

The risk engine combines forecast demand with estimated inventory and
safety-stock information.

This transforms a numerical forecast into an operational decision:

**Forecast → Inventory Gap → Risk Category → Priority → Recommended Action**

## 10. Financial Impact

The risk-scoring engine translates inventory conditions into estimated
financial exposure.

### Validated Financial Impact

| Financial Indicator | Estimated Impact |
|---|---:|
| Sales at Risk | ₹4,505,995 |
| Locked Inventory Capital | ₹9,475,885 |

### 1. Estimated Sales at Risk

**₹4,505,995**

This represents the estimated sales exposure associated with records
classified as **Stockout Risk**.

A stockout situation may prevent the business from fulfilling expected
demand, creating potential lost-sales exposure.

### 2. Estimated Locked Capital

**₹9,475,885**

This represents the estimated inventory capital associated with
records classified as **Overstock Risk**.

Excess inventory can reduce capital efficiency by keeping funds tied up
in products that may not be required immediately.

### Executive Interpretation

The analysis therefore identifies two competing financial pressures:

**Stockout Risk**
→ Potential lost sales

**Overstock Risk**
→ Capital tied up in inventory

The combined exposure highlights the value of balancing availability
with inventory efficiency rather than simply maximising stock levels.

### Important Note

These figures are **estimated exposure values produced by the
FORESIGHT risk-scoring methodology**. They should be interpreted as
decision-support indicators rather than confirmed realised losses or
savings.

## 11. Business Recommendations

Based on the forecasting and risk-scoring results, FORESIGHT recommends
a differentiated inventory strategy.

### Recommendation 1 — Prioritise Stockout Risk

**Priority: High**

There are **853,318 records** classified as Stockout Risk.

Recommended actions:

- Prioritise high-risk products for replenishment.
- Review available supplier and warehouse capacity.
- Monitor demand forecasts regularly.
- Focus first on products with the greatest estimated sales exposure.

**Decision: Reorder Immediately**

---

### Recommendation 2 — Reduce Overstock Exposure

**Priority: Medium**

There are **593,552 records** classified as Overstock Risk.

Recommended actions:

- Reduce unnecessary future purchasing.
- Identify slow-moving inventory.
- Consider targeted promotions.
- Review replenishment thresholds.
- Monitor inventory ageing.

**Decision: Reduce Purchasing / Launch Promotion**

---

### Recommendation 3 — Maintain Healthy Inventory

**Priority: Low**

There are **583,764 records** classified as Healthy.

Recommended actions:

- Maintain current inventory levels.
- Continue monitoring demand.
- Avoid unnecessary replenishment.
- Reassess risk as forecasts change.

**Decision: Maintain Current Inventory**

## 13. Prediction Service

FORESIGHT also includes a FastAPI-based prediction service.

### API Endpoints

| Endpoint | Purpose |
|---|---|
| `/health` | Service health check |
| `/predict/forecast` | Generate demand forecast |
| `/predict/risk` | Generate inventory risk classification |

### Forecast API

The forecasting endpoint accepts the 10 model features and returns a
predicted demand value.

### Risk API

The risk endpoint evaluates inventory conditions and returns:

- Risk category
- Priority
- Recommended decision
- Recommended action
- Inventory gap

→ **Demand Forecast**

→ **Risk Classification**
→ **Business Recommendation**

by applications beyond the Streamlit dashboard.
